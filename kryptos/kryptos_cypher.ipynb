```python
def first_row(alphabet, keyword):
    first_row = []
    keyword_reverse = keyword[::-1]

    # Insert reversed keyword characters into first_row list
    for letter_key in keyword_reverse:
        if letter_key not in first_row:
            first_row.insert(0, letter_key)

    # Append alphabet characters that are not in the keyword
    for letter in alphabet:
        if letter not in first_row:
            first_row.append(letter)

    return first_row

first_row = first_row(alphabet, keyword)
print(*first_row)
```

```python
def vigenere_table(first_row):
    # Initialize an empty grid to hold the Vigenère table
    grid = []

    # Create each row by cyclically shifting the first_row
    for c in range(len(first_row)):
        row = first_row[c:] + first_row[:c]
        grid.append(row)

    return grid

vigenere_table = vigenere_table(first_row)

for row in vigenere_table:
    print(' '.join(row))
```

```python
def encrypt(plain_text, key, vigenere_table):
    encrypt_text = ''
    filtered_text = plain_text.replace(" ", "")
    keystream = (key * (len(filtered_text) // len(key))) + key[:len(filtered_text) % len(key)]
    print('KEYSTREAM:', keystream)
    keystream_index = 0  # To keep track of the keystream position

    for c in range(len(plain_text)):
        if plain_text[c].isalpha():
            key_letter = keystream[keystream_index].upper()
            text_letter = plain_text[c].upper()
            keystream_index += 1
            
            idx_col = None
            if key_letter in vigenere_table[0]:
                idx_col = vigenere_table[0].index(key_letter)

            idx_row = None
            for i, row in enumerate(vigenere_table):
                if row[0] == text_letter:
                    idx_row = i
                    break

            encrypted_char = vigenere_table[idx_row][idx_col]

            if plain_text[c].isupper():
                encrypt_text += encrypted_char
            else:
                encrypt_text += encrypted_char.lower()

        else:
            encrypt_text += plain_text[c]

    return encrypt_text
```

```python
plain_text = "SECRET MESSAGE"
key = "HIDDEN"
print('PLAINTEXT:', plain_text)
encrypted_message = encrypt(plain_text, key, vigenere_table)
print('CIPHERTEXT:', encrypted_message)  # Expected: QKNEVW SKJJMSZ
```
